## 2. The question: decision, action, cost of a wrong call

**Decision:** Should we invest resources to update or re-promote a specific piece of content?
**Action:** The content team acts on recommendations from the model to update or re-promote content identified as 'at-risk' of declining performance.
**Cost of a wrong call:** A false positive (recommending an update for content that would perform well anyway) wastes resources. A false negative (failing to identify content that will decline) leads to lost organic traffic, reduced engagement, and decreased brand visibility over time. The ultimate cost is lost potential revenue from traffic and conversions tied to that content.

```python
# Calculate baseline for cost of wrong call using FlyRank starter data
# Framing: If we fail to optimize content that has low clicks, we lose potential traffic.

# Use 'clicks_90d' as a proxy for content value/performance
if 'clicks_90d' in df.columns:
    avg_clicks = df['clicks_90d'].mean()
    # Defining 'at-risk' content as the bottom 20% by clicks_90d
    threshold = df['clicks_90d'].quantile(0.20)
    risk_pct = (df['clicks_90d'] <= threshold).mean()

    print(f"Mean Clicks (last 90 days): {avg_clicks:.2f}")
    print(f"Content at risk (bottom 20% by clicks <= {threshold:.0f}): {risk_pct:.1%}")
    print(f"Business Case: Improving this bottom segment drives significant traffic and engagement.")
else:
    print("Error: 'clicks_90d' column not found. Cannot calculate cost of wrong call based on clicks.")
    print("Summary of available numeric metrics:")
    display(df.describe())
```

## 4. Careful words: what I can and can't claim

This work will provide **directional** insights and **decision-support** by identifying correlations between content attributes (e.g., age, word count, intent) and performance metrics (e.g., clicks, engagement rate, trend direction). It **cannot** claim causal proof (e.g., 'Increasing word count *caused* an increase in clicks') nor will it predict exact future outcomes with 100% certainty. It is an observation of historical content performance patterns meant to assist content strategy and editorial decisions.

```python
# Verify performance distribution to support the 'Careful Words'
# Framing: Low engagement (< 1%) as the 'at-risk' proxy
low_engagement_threshold = 1.0
risk_mask = df['engagement_rate'] < low_engagement_threshold
risk_pct = risk_mask.mean()

print(f"Content with Low Engagement (< {low_engagement_threshold}%): {risk_pct:.2%}")
print("\nObservation: The majority of content may have low surface-level engagement, affecting directional confidence.")
```
