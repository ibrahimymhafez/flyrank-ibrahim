## 1. My lane (or freestyle) and why

I am choosing the **E-commerce Customer Retention** lane. This lane is critical because acquiring new customers is significantly more expensive than retaining existing ones. By analyzing purchase frequency and churn patterns, I can identify which segments are at risk and recommend targeted re-engagement strategies.

## 2. The question: decision, action, cost of a wrong call

**Decision:** Should we issue a high-value discount coupon to a specific customer?
**Action:** The marketing team triggers an automated email campaign based on the model's churn probability score.
**Cost of a wrong call:** A false positive (predicting churn when they wouldn't) wastes marketing budget on unnecessary discounts. A false negative (missing a churning customer) results in the permanent loss of customer lifetime value (CLV).

```python
# Calculate baseline for cost of wrong call using FlyRank starter data
# Framing: If we fail to optimize content that has low conversions, we lose potential revenue.
if 'conversions' in df.columns:
    avg_conv = df['conversions'].mean()
    # Defining 'risk' as the bottom 20% of content by conversion rate
    threshold = df['conversions'].quantile(0.20)
    risk_pct = (df['conversions'] <= threshold).mean()

    print(f"Mean Conversions: {avg_conv:.2f}")
    print(f"Content at risk (<= {threshold} conversions): {risk_pct:.1%}")
    print(f"Business Case: Improving this bottom segment drives the primary success metric.")
else:
    print("Summary of available numeric metrics:")
    display(df.describe())
```

## 4. Careful words: what I can and can't claim

This work will provide **directional** insights and **decision-support** by identifying correlations between user behavior and churn. It **cannot** claim causal proof (e.g., 'Discount X *caused* Customer Y to stay') nor will it predict exact future outcomes with 100% certainty. It is an observation of historical patterns meant to assist marketing intuition.

```python
# Verify performance distribution to support the 'Careful Words'
# Framing: Low engagement (< 1%) as the 'at-risk' proxy
low_engagement_threshold = 1.0
risk_mask = df['engagement_rate'] < low_engagement_threshold
risk_pct = risk_mask.mean()

print(f"Content with Low Engagement (< {low_engagement_threshold}%): {risk_pct:.2%}")
print("\nObservation: The majority of content may have low surface-level engagement, affecting directional confidence.")
```
